```python
# Clean missing the Region
df["Region"]
```

```python
# Create State → Region mapping from existing data
state_to_region = (
    df.dropna(subset=["Region"]) # In this we drop the  missing value from the region column
      .groupby("State")["Region"] # In this we group them state and region 
      .agg(lambda x: x.mode().iloc[0]) # In this we create small function put the value in x and find mode 
      .to_dict() # In this we make the dictionary
)
```

```python
# To see the dictionary
state_to_region
```

```python
# To check the region
state_to_region["California"]
```

```python
# filling the missing data
df["Region"] = df["Region"].fillna(
    df["State"].map(state_to_region)
)
```

```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv( r"C:\Users\hp\OneDrive\Desktop\Ecommerce_sales Analysis\Ecommerce_Sales_clean.csv")
```

```python
df.head()
```

```python
df.head()
```

```python
df.columns.tolist()
```

```python
print("===== TOTAL KPIs =====")

print("Total Sales:", df["Sales"].sum())
print("Total Profit:", df["Profit"].sum())
print("Total Orders:", df["Order ID"].nunique())
print("Average Order Value:",
      df["Sales"].sum() / df["Order ID"].nunique())


print("\n===== SALES BY CATEGORY =====")

category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .reset_index()
      .sort_values("Sales", ascending=False)
)

display(category_sales)


print("\n===== SALES BY REGION =====")

region_sales = (
    df.groupby("Region")["Sales"]
      .sum()
      .reset_index()
      .sort_values("Sales", ascending=False)
)

display(region_sales)


print("\n===== TOP 10 PRODUCTS =====")

top_products = (
    df.groupby("Product")["Sales"]
      .sum()
      .reset_index()
      .sort_values("Sales", ascending=False)
      .head(10)
)

display(top_products)


print("\n===== PROFIT BY REGION =====")

region_profit = (
    df.groupby("Region")["Profit"]
      .sum()
      .reset_index()
      .sort_values("Profit", ascending=False)
)

display(region_profit)
```

```python
# Monthly sales 
monthly_sales = (
    df.groupby("Year Month")["Sales"]
      .sum()
      .reset_index()
)

monthly_sales
```

```python
#store by month
monthly_sales = monthly_sales.sort_values("Year Month")
```

```python
# chart 
plt.figure(figsize=(12, 5))

sns.lineplot(
    data=monthly_sales,
    x="Year Month",
    y="Sales",
    marker="o"
)

plt.title("Monthly Sales")
plt.xlabel("Year Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Sales by category
category_sales = (
    df.groupby("Category")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

category_sales
```

```python
# chart 
plt.figure(figsize=(8, 5))

sns.barplot(
    data=category_sales,
    x="Category",
    y="Sales"
)

plt.title("Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

```python
# profit by region
region_profit = (
    df.groupby("Region")["Profit"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

region_profit
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=region_profit,
    x="Region",
    y="Profit"
)

plt.title("Profit by Region")
plt.xlabel("Region")
plt.ylabel("Total Profit")
plt.tight_layout()
plt.show()
```

```python
# top 10 product 
top_products = (
    df.groupby("Product")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .head(10)
      .reset_index()
)

top_products
```

```python
# chart 
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_products,
    x="Sales",
    y="Product"
)

plt.title("Top 10 Products by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
```

```python
# sales by state 
state_sales = (
    df.groupby("State")["Sales"]
      .sum()
      .sort_values(ascending=False)
      .reset_index()
)

state_sales
```

```python
# chart 
top_states = state_sales.head(10)
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_states,
    x="Sales",
    y="State"
)

plt.title("Top 10 States by Sales")
plt.xlabel("Total Sales")
plt.ylabel("State")
plt.tight_layout()
plt.show()
```

```python
# sales vs profit 
plt.figure(figsize=(8, 6))

sns.scatterplot(
    data=df,
    x="Sales",
    y="Profit",
    alpha=0.6
)

plt.title("Sales vs Profit")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.tight_layout()
plt.show()
```

```python
df["Sales"].corr(df["Profit"])
```

```python
# discount vs profit
discount_profit = (
    df.groupby("Discount")["Profit"]
      .mean()
      .reset_index()
)

discount_profit
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    data=discount_profit,
    x="Discount",
    y="Profit"
)

plt.title("Average Profit by Discount")
plt.xlabel("Discount")
plt.ylabel("Average Profit")
plt.tight_layout()
plt.show()
```
